### Eigenvalues & Eigenvectors by Hand


```python
import numpy as np

A = np.array([
  [4 , 1 , 2],
  [1 , 5 , 1],
  [2 , 1 , 7]
], dtype = float)

eigenvalues , eigenvectors = np.linalg.eig(A)

for i in range(len(eigenvalues)):
  lam = eigenvalues[i]  #lambda in the formula
  v = eigenvectors[: , i]

  lhs = A @ v
  rhs = lam * v

  print(f"EigenValue {i}: matches = {np.allclose(lhs, rhs)}")

```

### Connect to PCA via the Iris Dataset

```python
from sklearn.datasets import load_iris
from sklearn.decomposition import PCA

# iris = load_data()
# X = iris.data
X = load_iris().data  # 150 samples, 4 features i.e. 150 rows and 4 columns


# Center the data (subtract the mean of each column)
X_centered = X - X.mean(axis = 0)

# Compute the covariance matrix by hand:
cov_matrix = (X_centered.T @ X_centered) / (X_centered.shape[0] - 1)

# Get its eigenvectors:
eigvals, eigvecs = np.linalg.eig(cov_matrix)

order = np.argsort(eigvals)[::-1]
eigvals_sorted = eigvals[order]
eigvecs_sorted = eigvecs[:, order]

pca = PCA(n_components=4)
pca.fit(X)

print(pca.components_)   # sklearn's principal components

#real because it prints +0.j which is imaginary part 
print(eigvecs_sorted.T.real)          # your manually sorted and computed ones

```
