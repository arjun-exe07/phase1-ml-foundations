```python
import numpy as np


def matmul_loop(A, B):
  m, n = A.shape
  a ,b = B.shape

  assert n == a, "Inner dimensions must match for matrix multiplication."

  C = np.zeros((m, b))

  for i in range(m):
    for j in range(b):
      for k in range(n):
        C[i, j] += A[i, k] * B[k, j]

  return C

  
```

```python
def matmul_vectorized(A, B):
    return A @ B  # or np.dot(A, B)
```

```python
import time

A = np.random.rand(200, 200)
B = np.random.rand(200, 200)

st = time.time()
C1 = matmul_loop(A, B)

print("Loop Time:", time.time() - st)

st = time.time()
C2 = matmul_vectorized(A, B)
print("Vectorized Time:", time.time() - st)
```
