### Part 1: Matrix multiplication vs vectorization

```python
import numpy as np


def matmul_loop(A, B):
  m, n = A.shape
  a ,b = B.shape

  assert n == a, "Inner dimensions must match for matrix multiplication."

  C = np.zeros((m, b))

  for i in range(m):
    for j in range(b):
      for k in range(n):
        C[i, j] += A[i, k] * B[k, j]

  return C


def matmul_vectorized(A, B):
    return A @ B  # or np.dot(A, B) 


import time

A = np.random.rand(200, 200)
B = np.random.rand(200, 200)

st = time.time()
C1 = matmul_loop(A, B)

print("Loop Time:", time.time() - st)

st = time.time()
C2 = matmul_vectorized(A, B)
print("Vectorized Time:", time.time() - st)
```

### Part 2: Matrix inversion
#### Guassian Elimination

```python
#NOt for where pivot is zero, but for the sake of simplicity, we will not implement pivoting here.

def guassian_inverse(A):
  n = A.shape[0]

  # Create an agumented matrix [A | I]
  I = np.eye(n)
  aug = np.hstack((A.astype(float) , I))

  for i in range(n):
    # Make the diagonal contain 1
    aug[i] = aug[i]/aug[i, i]

    for j in range(n):
      if i != j:
        aug[j] = aug[j] - aug[i]*aug[j, i]

  return aug[:, n:]  # Return the right half of the augmented matrix, which is the inverse


A = np.array([[4, 7, 2],
              [3, 6, 1],
              [2, 5, 3]], dtype=float)

my_inv = guassian_inverse(A)
numpy_inv = np.linalg.inv(A)

print(np.allclose(my_inv, numpy_inv))
```
